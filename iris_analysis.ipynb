Find the max and min petal length.

```python
print("Max:", df["petal_length"].max())
print("Min:", df["petal_length"].min())
```

Show flower types in a bar chart.

```python
import matplotlib.pyplot as plt

df["species"].value_counts().plot(kind="bar")

plt.title("Iris Flower Types")
plt.xlabel("Flower Type")
plt.ylabel("Count")
plt.show()
```

Compare petal length for each flower type.

```python
import matplotlib.pyplot as plt

df.boxplot(column="petal_length", by="species")

plt.title("Petal Length by Species")
plt.suptitle("")
plt.xlabel("Flower Type")
plt.ylabel("Petal Length")
plt.show()
```

Conclusion
